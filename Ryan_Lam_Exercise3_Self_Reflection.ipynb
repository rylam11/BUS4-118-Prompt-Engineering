# Exercise 3: Self-Reflection Prompt for Improving Output

## Goal
Use AI to generate a summary, critique its own response using specific criteria, and then produce an improved version.

## Tools Used
- Google Colab
- Python
- Gemini API

## Task
Generate a short summary explaining how artificial intelligence can be used in higher education, then use a self-reflection prompt to evaluate and improve the summary.

```python
!pip install -q -U google-genai

from google import genai
from google.colab import userdata
import time

api_key = userdata.get("GEMINI_API_KEY")
client = genai.Client(api_key=api_key)

def ask_gemini(prompt):
    for attempt in range(3):
        try:
            response = client.models.generate_content(
                model="gemini-3.5-flash-lite",
                contents=prompt
            )
            return response.text
        except Exception as e:
            print(f"Attempt {attempt + 1} failed: {e}")
            if attempt < 2:
                time.sleep(5)

    return "Unable to get a response after 3 attempts."

print("Gemini setup complete.")
```

## Original Summary

```python
original_prompt = """
Write a short summary explaining how artificial intelligence can be used in higher education.

Keep it between 120 and 150 words.
Write for a general college student audience.
Include examples of how AI can support students and universities.
"""

original_summary = ask_gemini(original_prompt)

print("ORIGINAL SUMMARY")
print(original_summary)
```

## Self-Reflection and Critique

```python
reflection_prompt = f"""
You are reviewing your own previous summary.

Original summary:
{original_summary}

Critique the summary using these criteria:

1. Accuracy - Is the information correct and reasonable?
2. Clarity - Is the summary easy for a general college student to understand?
3. Audience - Is the language appropriate for a general college student?
4. Length - Is the summary between 130 and 150 words?
5. Formatting - Is the summary organized and readable?

For each criterion:
- State what was done well.
- Identify anything that could be improved.

Then provide a short overall recommendation for how the summary should be revised.

Do not write the revised summary yet.
"""

reflection_output = ask_gemini(reflection_prompt)

print("SELF-REFLECTION AND CRITIQUE")
print(reflection_output)
```

## Revised Summary

```python
revision_prompt = f"""
Revise the original summary using the critique below.

Original summary:
{original_summary}

Critique:
{reflection_output}

Requirements for the revised summary:
- Keep the information accurate.
- Make the writing clear and easy for a general college student to understand.
- Keep the tone appropriate for a college student audience.
- Keep the summary between 130 and 150 words.
- Use clean, readable paragraph formatting.
- Preserve the strongest examples from the original summary.
- Improve any weaknesses identified in the critique.

Return only the revised summary.
"""

revised_summary = ask_gemini(revision_prompt)

print("REVISED SUMMARY")
print(revised_summary)
```

## Revision Note

The first revised summary was only 118 words, which did not meet the required 120–150 word range. I adjusted the prompt to target 130–150 words so the final response would stay safely within the required limit.

The final revised summary was 138 words.

```python
word_count = len(revised_summary.split())
print("Revised word count:", word_count)
```
